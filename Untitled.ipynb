# Introduction to vectors

1. Vectors:
   - Comuted relative to the positive X-axis
   - ordered list of numbers
   - quantity + direction
   - denoted as **a**
2. Dimensionality:
   - number of elements in a vector (shape or length)
3. Orientation:
   - Row orientation
   - Column orientation

## Python representation 

```python
import numpy as np
vectorAsList = [1,2,3,4]
vectorAsArray = np.array([1,2,3,4])

#row vector
rowVector = ([[1,2,3] ])

#column vector
colVector = ([ [1],[2],[3] ])
```

## Vector addition and subtraction

Only possible for vectors having same dimensionality

```python
a = np.array([1,2,3,4])
b = np.array([5,6,7,8])

print(a+b)
print(a-b)
```
